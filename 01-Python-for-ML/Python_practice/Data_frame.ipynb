```python
# Importing Dog CSV file - 
my_df3 = pd.read_csv('dog_data.csv')
```

```python
my_df3
```

```python
#count Disctinct Values - Decending 
my_df3['Color'].value_counts()
```

```python
#count Disctinct Values - Ascending 
my_df3['Color'].value_counts(ascending =True)
```

```python
#Dog Name - Without Nan
my_df3['DogName'].value_counts()
```

```python
#Dog Name - Nan
my_df3['DogName'].value_counts(dropna=False)
```

```python
# Get Relatice Frequency - Percentage
print(my_df3['Color'].value_counts(normalize=True))
```

```python
#Get Specific Item Count 
print(my_df3['Color'].value_counts()['WHITE'])
```
